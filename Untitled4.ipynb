```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python

class Course:
    def __init__(self, name):
        self.name = name
        self.assignments = []

    def add_assignment(self, name, category, score, total):
        assignment = {
            "name": name,
            "category": category,
            "score": score,
            "total": total
        }
        self.assignments.append(assignment)

    def calculate_average(self):
        earned_points = 0
        possible_points = 0

        for assignment in self.assignments:
            earned_points += assignment["score"]
            possible_points += assignment["total"]

        if possible_points == 0:
            return 0
        else:
            return (earned_points / possible_points) * 100
```

```python

    def show_assignments(self):
        print(f"\n--- Course: {self.name} ---")
        if len(self.assignments) == 0:
            print("No assignments recorded yet.")
            return

        for assignment in self.assignments:
            print("Assignment:", assignment["name"])
            print("Category:", assignment["category"])
            print("Score:", assignment["score"], "/", assignment["total"])
            print()
```

Partner 1

```python
courses = {}
study_hours = {}

def add_course():
    name = input("Enter course name: ").strip()

    if name == "":
        print("Please enter a course name.")
    elif name in courses:
        print("This course already exists.")
    else:
        courses[name] = Course(name)
        print(f"Course '{name}' added successfully!")

def record_assignment():
    name = input("Enter course name: ").strip()

    if name in courses:
        assignment_name = input("Enter assignment name: ").strip()
        category = input("Enter category (Exam, Quiz, Homework): ").strip()

        try:
            score = float(input("Enter points earned: "))
            total = float(input("Enter total possible points: "))

            if total > 0 and 0 <= score <= total:
                courses[name].add_assignment(assignment_name, category, score, total)
                print("Assignment recorded successfully!")
            else:
                print("Invalid score or total points.")
        except ValueError:
            print("Please enter valid numbers.")
    else:
        print("Course not found.")

def academic_standing():
    if len(courses) == 0:
        print("No courses added yet.")
        return

    for name in courses:
        if len(courses[name].assignments) == 0:
            print(name, "- No grades recorded")
        else:
            average = courses[name].calculate_average()

            if average >= 90:
                standing = "Excellent"
            elif average >= 80:
                standing = "Good"
            elif average >= 70:
                standing = "Satisfactory"
            else:
                standing = "Needs Improvement"

            print(f"{name} - {round(average, 2)}% - {standing}")

def remove_course():
    name = input("Enter course name to remove: ").strip()

    if name in courses:
        del courses[name]
        if name in study_hours:
            del study_hours[name]
        print("Course removed successfully!")
    else:
        print("Course not found.")
```

Partner 2

```python
def record_study_hours():
    name = input("Enter course name: ").strip()

    if name in courses:
        topic = input("What did you study? ").strip()
        date = input("Enter date (MM/DD/YYYY): ").strip()

        try:
            hours = float(input("How many hours did you study? "))

            if hours > 0:
                if name not in study_hours:
                    study_hours[name] = []

                session = {
                    "topic": topic,
                    "date": date,
                    "hours": hours
                }
                study_hours[name].append(session)
                print("Study hours recorded successfully!")
            else:
                print("Hours must be greater than zero.")
        except ValueError:
            print("Please enter a valid number.")
    else:
        print("Course not found.")

def calculate_study_hours():
    if len(courses) == 0:
        print("No courses added yet.")
        return

    for name in courses:
        total_hours = sum(session["hours"] for session in study_hours.get(name, []))
        print(f"{name} - {total_hours} study hours")

def show_study_sessions():
    name = input("Enter course name: ").strip()

    if name in courses:
        if name in study_hours and len(study_hours[name]) > 0:
            print(f"\n--- Study Sessions for {name} ---")
            for session in study_hours[name]:
                print("Topic:", session["topic"])
                print("Date:", session["date"])
                print("Hours:", session["hours"])
                print()
        else:
            print("No study sessions recorded.")
    else:
        print("Course not found.")

def create_academic_report():
    report = []

    for name in courses:
        if len(courses[name].assignments) > 0:
            average = courses[name].calculate_average()
        else:
            average = None

        total_hours = sum(session["hours"] for session in study_hours.get(name, []))
        number_sessions = len(study_hours.get(name, []))

        course_info = {
            "Course": name,
            "Average Grade": average,
            "Study Hours": total_hours,
            "Study Sessions": number_sessions,
            "Assignments": len(courses[name].assignments)
        }
        report.append(course_info)

    report_df = pd.DataFrame(report)

    print("\n=== ACADEMIC PERFORMANCE REPORT ===")
    print(report_df)

    if len(report_df) > 0:
        print("\n=== SUMMARY STATISTICS ===")
        print(report_df.describe())
        print("\nTOTAL STUDY HOURS:", report_df["Study Hours"].sum())
        print("AVERAGE STUDY HOURS PER COURSE:", round(report_df["Study Hours"].mean(), 2))
        print("AVERAGE GRADE ACROSS COURSES:", round(report_df["Average Grade"].mean(), 2))

    return report_df

def show_graphs():
    report_df = create_academic_report()

    if len(report_df) == 0:
        print("No courses available.")
        return

    report_df.plot(
        kind="bar",
        x="Course",
        y="Study Hours",
        title="Study Hours by Course",
        legend=False
    )
    plt.ylabel("Hours")
    plt.show()

    graded_df = report_df.dropna(subset=["Average Grade"])

    if len(graded_df) > 0:
        graded_df.plot(
            kind="bar",
            x="Course",
            y="Average Grade",
            title="Average Grade by Course",
            legend=False
        )
        plt.ylabel("Grade (%)")
        plt.show()

        graded_df.plot(
            kind="scatter",
            x="Study Hours",
            y="Average Grade",
            title="Study Hours vs Average Grade"
        )
        plt.show()
    else:
        print("No grades recorded yet for grade charts.")
```

Main Menu

```python
def main():
    while True:
        print("\n======================================")
        print("STUDENT ACADEMIC PERFORMANCE TRACKER")
        print("======================================")
        print("1. Add a course")
        print("2. Record an assignment")
        print("3. View course assignments")
        print("4. View course averages")
        print("5. Check academic standing")
        print("6. Remove a course")
        print("7. Record study hours")
        print("8. View study sessions")
        print("9. View total study hours")
        print("10. Generate Pandas report")
        print("11. View performance graphs")
        print("12. Exit")

        choice = input("Choose an option (1-12): ").strip()

        if choice == "1":
            add_course()
        elif choice == "2":
            record_assignment()
        elif choice == "3":
            name = input("Enter course name: ").strip()
            if name in courses:
                courses[name].show_assignments()
            else:
                print("Course not found.")
        elif choice == "4":
            for name in courses:
                if len(courses[name].assignments) == 0:
                    print(name, "- No grades recorded")
                else:
                    average = courses[name].calculate_average()
                    print(f"{name} - {round(average, 2)}%")
        elif choice == "5":
            academic_standing()
        elif choice == "6":
            remove_course()
        elif choice == "7":
            record_study_hours()
        elif choice == "8":
            show_study_sessions()
        elif choice == "9":
            calculate_study_hours()
        elif choice == "10":
            create_academic_report()
        elif choice == "11":
            show_graphs()
        elif choice == "12":
            print("Thank you for using the tracker!")
            break
        else:
            print("Invalid option. Please try again.")
```

```python
main()
```
